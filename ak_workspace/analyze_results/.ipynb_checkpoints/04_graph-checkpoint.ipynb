```python
cell_types = ['Oligo_NN', 'OPC_NN', 'Astro-TE_NN', 'L23_IT_CTX_Glut', 'L6_IT_CTX_Glut', 'L6_CT_CTX_Glut', 'L5_ET_CTX_Glut', 'CA1-ProS_Glut', 'CA3_Glut', 'DG_Glut']
```

## Correlations

```python
CT_RES = {}
for i, _ct in enumerate(cell_types): 
    
    RES = {}
    for _deg, _int_deg in zip(['down', 'non', 'up'], [0, 1, 2]): 
        # df_shap = pd.read_csv(f"../result_data/xgb_mods_2/shap_feats/{_ct}.cv{extension}.shap_{_deg}.csv", index_col=0)
        df_shap = pd.read_csv(f"../result_data/xgb_mods_2/shap_feats/{_ct}.bal_det.shap_{_deg}.csv", index_col=0)
        df_shap = df_shap[(df_shap['DEG'] == df_shap['pred']) & (df_shap['DEG'] == _int_deg)]
        
        stat = df_shap[['DEG', 'pred']]
        df_shap = df_shap[df_shap.columns.difference(['DEG', 'pred'])].T
        gene_list = df_shap.columns
        
        df_shap['modality'] = [idx.split(".")[-1] for idx in df_shap.index]
        df_shap['modality'] = df_shap['modality'].map(FEATURE_MAP)
        # break
        df_shap = df_shap.groupby("modality").sum().abs()
        # df_shap_abs = df_shap.copy()
        # df_shap_abs[gene_list] = df_shap[gene_list].abs()
        
        # df_slice_list = []
        # for i, _gene in enumerate(gene_list): 
        #     indices = df_shap_abs[[_gene, 'modality']].groupby("modality")[_gene].nlargest(N).index.get_level_values(1)
        #     # df_slice = df_shap_abs.loc[indices, [_gene, 'modality']].groupby("modality").mean()
        #     df_slice = df_shap.loc[indices, [_gene, 'modality']].groupby("modality").sum().abs()
        #     df_slice_list.append(df_slice)
        # df_shap = pd.concat(df_slice_list, axis=1)
        
        
        df_list = []
        for m in FEAT_ORDER: 
            df_temp = df_shap[df_shap.index == m]
            # df_temp = df_temp.drop(columns = ['modality'])
            M = df_temp.shape[0]
            if M == 0: 
                df_temp = pd.DataFrame(df_shap.columns).set_index("gene_name")
                df_temp[m] = 0
                df_list.append(df_temp)
            else: 
                df_list.append(df_temp.T)
                # df_list.append(pd.DataFrame(df_temp).rename(columns={0:m}))
        
        df_shap = pd.concat(df_list, axis=1).T 
        # df_shap = df_shap.groupby('modality').mean() # .mean(axis=1)
        df_shap = df_shap.apply(lambda x: (x - min(x)) / sum(x - min(x)), axis=0)
        RES[_deg] = df_shap.mean(axis=1)
    
    CT_RES[_ct] = RES
```

```python
df_list = []
for _ct, shap_dict in CT_RES.items(): 
    df_mod = pd.DataFrame.from_dict(shap_dict).reset_index(names="index").melt(id_vars='index', value_name=_ct)
    df_mod['deg_modality'] = df_mod['variable'] + '_' + df_mod['index']
    df_mod = df_mod.drop(columns = ['variable', 'index']).set_index('deg_modality')
    df_list.append(df_mod)
df_mod = pd.concat(df_list, axis=1)

pdist = spc.distance.pdist(df_mod.T)
linkage = spc.linkage(pdist, method='complete')
idx = spc.fcluster(linkage, 0.6*pdist.max(), 'distance')
columns = [df_mod.columns.tolist()[i] for i in list(np.argsort(idx))]
```

```python
corr_vals = df_mod[columns].corr()
corr_vals = np.power(corr_vals, 2)

# sns.heatmap(corr_vals, vmin = 0, vmax = 1)

fig, ax = plt.subplots(1,1,figsize=(6, 5), dpi=300, constrained_layout=True)

g = ax.imshow(corr_vals, vmin=0, vmax=1, cmap="Reds")

ax.set_xticks(range(corr_vals.shape[1]), labels=corr_vals.columns, 
                  rotation=45, ha='right', rotation_mode='anchor')
ax.set_yticks(range(corr_vals.shape[0]), labels = corr_vals.index)


sns.despine(bottom=True, left=True)


fig.colorbar(g, ax=ax, ticks=[0, 0.5, 1], label="", location='right', aspect=10, shrink=0.25, fraction=1)

# plt.savefig("../images/fig6/celltype_corr.pdf", bbox_inches= 'tight')
plt.show()
plt.close()
```
